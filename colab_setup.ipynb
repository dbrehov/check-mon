# 🛠 Пошаговая установка окружения для check-mon
Запускайте ячейки по очереди. Если пакет уже установлен, установка будет пропущена.

```python
%%capture
print("Checking Node.js...")
!node -v >/dev/null 2>&1 || (curl -fsSL https://deb.nodesource.com/setup_24.x | sudo -E bash - && sudo apt-get install -y nodejs)
!node -v
```

```python
%%capture
print("Checking system dependencies (Linux libs)...")
# Проверяем наличие xvfb как индикатор установки системных библиотек
!dpkg -l | grep xvfb >/dev/null 2>&1 || (sudo apt-get update && sudo apt-get install -y libnss3 libatk1.0-0 libatk-bridge2.0-0 libcups2 libxkbcommon0 libxdamage1 libxcomposite1 libxrandr2 libxfixes3 libxext6 libx11-6 libglib2.0-0 libpango-1.0-0 libcairo2 libasound2 libatk1.0-data fonts-liberation wget xvfb)
print("System dependencies check complete.")
```

```python
%%capture
print("Checking global tools (typescript, ts-node)...")
!tsc -v >/dev/null 2>&1 || sudo npm install -g typescript ts-node
!tsc -v
```

```python
%%capture
print("Checking Playwright Chromium...")
!npx playwright --version >/dev/null 2>&1 || npx playwright install chromium
!npx playwright --version
```

```python
%%capture
print("Checking yt-dlp...")
!yt-dlp --version >/dev/null 2>&1 || pip install yt-dlp
!yt-dlp --version
```
